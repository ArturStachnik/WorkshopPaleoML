## Paso 6. Cargar un CSV

Un CSV es una tabla guardada como texto, con los valores separados por comas. Se puede abrir con Excel.

Este tiene 300 muestras de sedimento del fondo del mar. Para cada una indica dónde se tomó, la temperatura del agua en superficie y el porcentaje de tres especies de foraminíferos planctónicos.

**Qué hacer:** ejecutar la celda.

**Qué debe aparecer:** las cinco primeras filas de la tabla, como en una hoja de cálculo.

```python
import os

URL = ("https://raw.githubusercontent.com/ArturStachnik/WorkshopPaleoML/"
       "main/datos/demo_colab.csv")

# si se trabaja con el repositorio descargado, se lee del disco
LOCAL = "../datos/demo_colab.csv"
datos = pd.read_csv(LOCAL if os.path.exists(LOCAL) else URL)

datos.head()
```

## Paso 8. Una figura

**Qué hacer:** ejecutar la celda.

**Qué debe aparecer:** cada punto es una muestra. En azul, *N. pachyderma*; en rojo, *G. ruber*.

Una especie domina en aguas frías y la otra en aguas cálidas. Esa relación entre especies y temperatura es la que se convierte en un modelo en el bloque 2.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.scatter(datos["temperatura_agua"], datos["N_pachyderma"], s=14,
           color="#1f6e8c", label="N. pachyderma")
ax.scatter(datos["temperatura_agua"], datos["G_ruber"], s=14,
           color="#c2462e", label="G. ruber")
ax.set_xlabel("Temperatura del agua en superficie (°C)")
ax.set_ylabel("Abundancia en la muestra (%)")
ax.legend()
plt.show()
```
